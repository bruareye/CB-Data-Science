## Objetivo

## ¿Qué es apply?
`apply()` es un método que permite aplicar una función a cada elemento de una Serie (columna).

## Parte 1: Repaso de Funciones en Python

Antes de usar `apply()`, es importante entender cómo crear funciones.

### Estructura de una Función:
```python
def nombre_funcion(parametro1, parametro2):
    # Código de la función
    resultado = parametro1 + parametro2
    return resultado

# Llamar la función
nombre_funcion(5, 3)  # Retorna 8
```

### Componentes:
- `def`: palabra clave para definir función
- `nombre_funcion`: nombre descriptivo
- `parametros`: valores que recibe la función
- `return`: valor que devuelve la función

### Ejemplos Prácticos:

````python
# Ejemplo 1: Función simple que calcula el doble de un número
def calcular_doble(numero):
    return numero * 2

# Probar la función
print(calcular_doble(5))   # 10
print(calcular_doble(3.5)) # 7.0
````

````python
# Ejemplo 2: Función que calcula porcentaje de propina
def calcular_porcentaje_propina(total, propina):
    porcentaje = (propina / total) * 100
    return porcentaje

# Probar
resultado = calcular_porcentaje_propina(100, 15)
print('Porcentaje:', resultado, '%')
````

````python
# Ejemplo 3: Función que categoriza montos
def categorizar_cuenta(monto):
    if monto < 15:
        return 'Bajo'
    elif monto < 25:
        return 'Medio'
    else:
        return 'Alto'

# Probar
print(categorizar_cuenta(10))  # Bajo
print(categorizar_cuenta(20))  # Medio
print(categorizar_cuenta(30))  # Alto
````

## Parte 2: Funciones Lambda

Las funciones lambda son funciones anónimas de una sola línea.

### Estructura:
```python
# Función normal
def sumar(x, y):
    return x + y

# Función lambda equivalente
sumar = lambda x, y: x + y
```

### Cuándo usar Lambda:
- Funciones simples de una sola línea
- Uso temporal (no necesitas reutilizarla)
- Dentro de funciones como `apply()`

### Ejemplos:

````python
# Ejemplo 1: Lambda simple
doble = lambda x: x * 2
print(doble(5))  # 10
````

````python
# Ejemplo 2: Lambda con múltiples parámetros
suma = lambda x, y: x + y
print(suma(3, 7))  # 10
````

````python
# Ejemplo 3: Lambda con condicional simple
es_mayor = lambda edad: 'Mayor' if edad >= 18 else 'Menor'
print(es_mayor(20))  # Mayor
print(es_mayor(15))  # Menor
````

````python
# Ejemplo 4: Comparación función normal vs lambda

# Función normal
def calcular_propina_normal(total):
    return total * 0.15

# Lambda equivalente
calcular_propina_lambda = lambda total: total * 0.15

# Ambas funcionan igual
print(calcular_propina_normal(100))
print(calcular_propina_lambda(100))
````

## Parte 3: Apply en Series

Aplica una función a cada elemento de una columna (Serie).

### Estructura:
```python
# Con función definida
def mi_funcion(valor):
    return valor * 2

df['columna'].apply(mi_funcion)

# Con lambda
df['columna'].apply(lambda x: x * 2)

# Asignar a nueva columna
df['nueva_columna'] = df['columna'].apply(mi_funcion)
```

### Ejemplos:

````python
# Ejemplo 1: Multiplicar cada propina por 100
def multiplicar_por_100(valor):
    return valor * 100

# Aplicar a la columna tip
tips['tip'].apply(multiplicar_por_100).head()
````

````python
# Ejemplo 2: Categorizar total_bill usando función
def categorizar_cuenta(monto):
    if monto < 15:
        return 'Económico'
    elif monto < 30:
        return 'Moderado'
    else:
        return 'Costoso'

tips['categoria_cuenta'] = tips['total_bill'].apply(categorizar_cuenta)
tips[['total_bill', 'categoria_cuenta']].head(10)
````

````python
# Ejemplo 3: Usar lambda directamente
tips['tip_redondeado'] = tips['tip'].apply(lambda x: round(x))
tips[['tip', 'tip_redondeado']].head()
````

````python
# Ejemplo 4: Convertir texto a mayúsculas
tips['day_mayusculas'] = tips['day'].apply(lambda x: str(x).upper())
tips[['day', 'day_mayusculas']].head()
````
