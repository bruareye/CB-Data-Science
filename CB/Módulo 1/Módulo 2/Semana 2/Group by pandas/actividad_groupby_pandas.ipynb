## Parte 3: Introducción a .agg()

El método `.agg()` permite mayor flexibilidad para especificar funciones de agregación.

### Estructura General:
```python
# Forma 1: Pasar el nombre de la función como string
df.groupby('columna')['columna_numerica'].agg('mean')
df.groupby('columna')['columna_numerica'].agg('sum')

# Forma 2: Usar funciones de numpy
import numpy as np
df.groupby('columna')['columna_numerica'].agg(np.mean)
df.groupby('columna')['columna_numerica'].agg(np.median)

# Forma 3: Múltiples funciones (retorna DataFrame)
df.groupby('columna')['columna_numerica'].agg(['min', 'max'])
```

## Parte 6: Introducción a .unstack()

`.unstack()` transforma un nivel del índice en columnas, creando tablas más legibles.

### Estructura General:
```python
# Primero: crear un groupby con índice múltiple
resultado = df.groupby(['columna1', 'columna2'])['columna_numerica'].mean()
# Resultado: Series con MultiIndex

# Aplicar unstack() - convierte el último nivel del índice en columnas
resultado.unstack()

# Unstack de un nivel específico
resultado.unstack(level=0)  # Convierte el primer nivel en columnas
resultado.unstack(level=1)  # Convierte el segundo nivel en columnas

# Unstack con fill_value para reemplazar NaN
resultado.unstack(fill_value=0)

# Unstack de múltiples niveles
resultado.unstack(level=[0, 1])
```

**Resultado:** Tabla pivote con una dimensión como filas y otra como columnas

## Resumen de Funciones Útiles

### Funciones de agregación:
- `mean()`, `median()`, `sum()`, `count()`, `size()`
- `min()`, `max()`, `std()`, `var()`
- `first()`, `last()`, `nunique()`

### Métodos importantes:
- `.groupby()` - agrupar datos
- `.agg()` - aplicar funciones de agregación
- `.unstack()` - pivotar índice a columnas
- `.reset_index()` - convertir índice en columna
- `.transform()` - aplicar función y mantener forma original
- `.describe()` - estadísticas descriptivas completas
