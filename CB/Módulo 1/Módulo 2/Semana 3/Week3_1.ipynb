## Parte 3: Análisis de Clientes

### Creación del dataset de clientes
Vamos a agregar los datos a nivel de cliente para poder hacer segmentación y análisis RFM (Recency, Frequency, Monetary).

```python
# Agregamos los datos por cliente para obtener métricas clave
df_clients = df.groupby("customer_id", as_index = False).agg({
    "invoice_id": "nunique",      # Número de facturas (frecuencia de compra)
    "sku": "nunique",             # Variedad de productos comprados
    "amount": "sum",              # Valor total de compras (valor monetario)
    "purchase_datetime": "max"    # Última fecha de compra (recency)
})

# Calculamos el ticket promedio por cliente
df_clients["ticket_promedio"] = df_clients["amount"] / df_clients["invoice_id"]

# Renombramos las columnas para mayor claridad
df_clients.columns = ["customer_id", "num_facturas", "skus", "compras", "ultima_compra", "ticket_promedio"]
df_clients.head(2)
```
