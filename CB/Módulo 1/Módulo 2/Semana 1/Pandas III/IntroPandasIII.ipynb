## Convertir la fecha en formato fecha (Paso pendiente)
hint: vamos a utilizar la función pd.to_datetime().  
format = "mixed" en caso que el formato de la columa de fechas no sea el mismo

```python
df["fecha_registro"] = pd.to_datetime(df["fecha_registro"], format = "mixed")
```

```python
df["fecha_registro"].dt.day_name()
```

## to_numeric

```python
# pd.to_numeric convierte valores a números y permite manejar errores.
# errors="coerce" convierte los valores no numéricos en NaN, en lugar de generar un error.
edad_numerica = pd.to_numeric(df["edad"], errors="coerce")
display(edad_numerica)
```

```python
# Ejemplo con una serie que incluye texto no convertible:
ejemplo = pd.Series(["25", "30", "no disponible"])
pd.to_numeric(ejemplo, errors="coerce")

# Diferencias frente a astype:
# - pd.to_numeric está diseñada específicamente para conversiones numéricas.
# - pd.to_numeric permite usar errors="coerce" para reemplazar errores por NaN,
#   o errors="raise" para detener la ejecución si encuentra valores inválidos.
# - astype convierte a un tipo de dato específico, por ejemplo .astype(float),
#   pero normalmente genera un error si existen textos no convertibles
```
