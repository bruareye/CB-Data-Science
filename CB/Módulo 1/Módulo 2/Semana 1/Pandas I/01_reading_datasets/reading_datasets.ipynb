## Summary: Key Parameters for Reading Datasets

### For CSV files (`pd.read_csv()`):
- `encoding`: Specify character encoding (`'utf-8'`, `'latin-1'`, `'iso-8859-1'`)
- `sep` or `delimiter`: Specify column separator (`,`, `;`, `|`, `\t`)
- `skiprows`: Number of rows to skip at the beginning
- `index_col`: Column to use as row index
- `header`: Row number to use as column names (default: 0)

### For Excel files (`pd.read_excel()`):
- `sheet_name`: Name or index of sheet to read
- `skiprows`: Rows to skip
- `usecols`: Columns to read

### For Parquet files (`pd.read_parquet()`):
- Usually no parameters needed!
- Encoding and data types are automatically preserved
