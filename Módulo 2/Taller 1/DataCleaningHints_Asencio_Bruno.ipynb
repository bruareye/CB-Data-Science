## 3) Limpieza columnas de texto
Identifique problemas en las columnas de texto, y corríjalas. Guarde los cambios en el mismo dataframe

<details>
<summary><b>💡 Hints</b></summary>

- **Sexo**: Hay valores como 'female', 'm', 'M', 'F'. Estandariza a solo 'M' y 'F'

```python
# Tu código aquí
df["sexo"].unique()
```

```python
df["ciudad"].unique()
```
