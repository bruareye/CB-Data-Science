### Métodos de Segmentación - ¿Cuándo usar cada uno?

| Método | Cuándo usarlo | Ejemplo |
|--------|---------------|----------|
| `apply()` | Lógica compleja o condiciones múltiples | Clasificar por múltiples criterios |
| `pd.cut()` | Rangos de negocio conocidos | Clasificar por ingreso: bajo, medio, alto |
| `pd.qcut()` | Distribución equitativa | Dividir clientes en 4 grupos iguales |
| Pareto | Identificar los más importantes | El 20% que genera el 80% |
